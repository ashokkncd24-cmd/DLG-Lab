```python
import numpy as np
def sigmoid(x):
    return 1/(1+np.exp(-x))
def sigmoid_derivative(a):
    return a*(1-a)

inputs = np.array([[0,0],[0,1],[1,0],[1,1]])
expected_output = np.array([[0], [1], [1], [0]])
inN,hidN,outN=2,2,1
hidden_wt=np.random.uniform(size=(inN,hidN))
hidden_bias=np.random.uniform(size=(1,hidN))
output_wt=np.random.uniform(size=(hidN,outN))
output_bias=np.random.uniform(size=(1,outN))
lr=0.1

for epoch in range(10000):
    h_out=sigmoid(np.dot(inputs,hidden_wt)+hidden_bias)
    predicted_output = sigmoid(np.dot(h_out, output_wt) + output_bias)

    error=expected_output-predicted_output
    d_out=error*sigmoid_derivative(predicted_output)
    d_hid=d_out.dot(output_wt.T) * sigmoid_derivative(h_out)

    output_wt += h_out.T.dot(d_out) * lr
    output_bias += np.sum(d_out, axis=0, keepdims=True) * lr
    hidden_wt += inputs.T.dot(d_hid) * lr
    hidden_bias +=np.sum(d_hid,axis=0,keepdims=True)*lr

print(np.round(predicted_output,3))
```

```python
#using keras
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD
X=np.array([[0,0],[0,1],[1,0],[1,1]])
Y = np.array([[0], [1], [1], [0]])
model=Sequential()
model.add(Dense(8, input_dim=2, activation='relu'))
model.add(Dense(1, activation='sigmoid'))
model.compile(loss='binary_crossentropy',optimizer=SGD(learning_rate=0.1), metrics=['accuracy'] )
model.fit(X, Y, epochs=1000, verbose=0)
_, acc = model.evaluate(X, Y, verbose=0)
print(f"Accuracy: {acc*100:.2f}%")
print(np.round(model.predict(X)))
```

Result

The neural network successfully learned the XOR pattern using backpropagation. The predicted outputs are close to the expected outputs:

* 0 XOR 0 = 0
* 0 XOR 1 = 1
* 1 XOR 0 = 1
* 1 XOR 1 = 0
